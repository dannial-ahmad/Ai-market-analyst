2.3 Fix 2: standardize categorical columns

Technique: lowercase each value, then look it up in a dictionary that sends every variant to one official spelling. The helper reports anything it couldn't map, so nothing slips through silently.

```python
def standardize(series, mapping, col_name):
    key = series.astype(str).str.strip().str.lower()
    out = key.map(mapping)
    out[series.isna()] = np.nan                      # keep real missing values missing
    unmapped = series[out.isna() & series.notna()].unique()
    print(f"{col_name:22s} -> {out.nunique():2d} clean values | unmapped: {list(unmapped)}")
    return out

MAPS = {
 'country': {'pakistan': 'Pakistan', 'pk': 'Pakistan'},
 'work_type': {'remote': 'Remote', 'work from home': 'Remote', 'wfh': 'Remote', 'hybrid': 'Hybrid',
               'on-site': 'On-site', 'onsite': 'On-site', 'on site': 'On-site'},
 'employment_type': {'full-time': 'Full-time', 'full time': 'Full-time', 'fulltime': 'Full-time',
                     'part-time': 'Part-time', 'part time': 'Part-time',
                     'contract': 'Contract', 'contractual': 'Contract',
                     'internship': 'Internship', 'intern': 'Internship'},
 'experience_level': {'internship': 'Internship', 'intern': 'Internship',
                      'entry level': 'Entry Level', 'entry-level': 'Entry Level', 'entry': 'Entry Level',
                      'junior': 'Junior', 'jr': 'Junior', 'junior level': 'Junior',
                      'mid level': 'Mid Level', 'mid-level': 'Mid Level', 'intermediate': 'Mid Level',
                      'senior': 'Senior', 'sr': 'Senior', 'senior level': 'Senior'},
 'education_required': {"bachelor's": "Bachelor's", 'bachelors': "Bachelor's", "bachelor's degree": "Bachelor's",
                        'bachelor': "Bachelor's", 'bs': "Bachelor's",
                        "master's": "Master's", 'masters': "Master's", "master's degree": "Master's",
                        'ms': "Master's", 'msc': "Master's",
                        'phd': 'PhD', 'ph.d': 'PhD', 'doctorate': 'PhD',
                        "bachelor's or master's": "Bachelor's or Master's", "bachelor's/master's": "Bachelor's or Master's",
                        'bs or ms': "Bachelor's or Master's", 'bachelors or masters': "Bachelor's or Master's",
                        'no degree required': 'No Degree Required', 'not required': 'No Degree Required'},
 'industry': {'technology': 'Technology', 'it': 'Technology', 'information technology': 'Technology', 'tech': 'Technology',
              'banking': 'Banking', 'bank': 'Banking',
              'finance': 'Finance', 'financial services': 'Finance', 'fintech': 'Finance',
              'e-commerce': 'E-commerce', 'ecommerce': 'E-commerce',
              'healthcare': 'Healthcare', 'health care': 'Healthcare', 'health': 'Healthcare',
              'telecom': 'Telecom', 'telecommunications': 'Telecom',
              'education': 'Education', 'edtech': 'Education',
              'manufacturing': 'Manufacturing', 'retail': 'Retail',
              'government': 'Government', 'govt': 'Government', 'public sector': 'Government',
              'logistics': 'Logistics', 'supply chain': 'Logistics',
              'energy': 'Energy', 'power & energy': 'Energy'},
 'company_size': {'startup': 'Startup', 'start-up': 'Startup', 'small': 'Small', 'medium': 'Medium',
                  'mid-size': 'Medium', 'large': 'Large', 'enterprise': 'Enterprise'},
 'job_source': {'linkedin': 'LinkedIn', 'indeed': 'Indeed', 'glassdoor': 'Glassdoor',
                'company website': 'Company Website', 'company site': 'Company Website',
                'university career portal': 'University Career Portal', 'univ career portal': 'University Career Portal',
                'recruitment agency': 'Recruitment Agency', 'agency': 'Recruitment Agency'},
 'visa_sponsorship': {'yes': 'Yes', 'y': 'Yes', 'no': 'No', 'n': 'No', 'not specified': 'Not Specified', 'n/a': 'Not Specified'},
 'target_role_category': {'data analytics': 'Data Analytics', 'data science': 'Data Science',
                          'machine learning': 'Machine Learning', 'artificial intelligence': 'Artificial Intelligence',
                          'data engineering': 'Data Engineering', 'business intelligence': 'Business Intelligence',
                          'software development': 'Software Development'},
}

for col, mapping in MAPS.items():
    df_clean[col] = standardize(df_clean[col], mapping, col)

# currency: simple upper-case (pkr -> PKR)
df_clean['currency'] = df_clean['currency'].str.upper()

# remote_allowed: mixed True/False/'Yes'/'No' -> real boolean
ra_map = {'true': True, 'yes': True, 'false': False, 'no': False}
df_clean['remote_allowed'] = df_clean['remote_allowed'].astype(str).str.lower().map(ra_map)
```

2.3b Location, company name and job title

```python
# ---------- LOCATION: keep the city only, or 'Remote' ----------
def clean_location(x):
    if pd.isna(x):
        return x
    if 'remote' in x.lower():
        return 'Remote'
    return x.split(',')[0].strip().title()      # 'lahore, Punjab' -> 'Lahore'

df_clean['location'] = df_clean['location'].apply(clean_location)

# ---------- COMPANY: remove legal suffixes, unify capitalization ----------
legal_suffix = r'\s*(\(pvt\.?\)|pvt\.?|private)?\s*(ltd\.?|limited)\s*$'
company_base = df_clean['company'].str.replace(legal_suffix, '', case=False, regex=True).str.strip()

# For each company, keep its most common capitalization ('NEXORA DATA' -> 'Nexora Data')
canonical = company_base.groupby(company_base.str.lower()).agg(lambda s: s.value_counts().idxmax())
df_clean['company'] = company_base.str.lower().map(canonical)

# ---------- JOB TITLE: expand abbreviations, then map to the official list ----------
BASE_TITLES = ['Data Analyst', 'Data Scientist', 'Junior Data Scientist', 'Machine Learning Engineer', 'AI Engineer',
               'BI Analyst', 'Business Analyst', 'Data Engineer', 'Python Developer',
               'ML Intern', 'Data Science Intern', 'Data Analyst Intern']
PREFIXES = ['', 'Senior ', 'Lead ', 'Junior ', 'Associate ']
official_titles = {(p + b).lower(): (p + b) for p in PREFIXES for b in BASE_TITLES}

def clean_title(t):
    if pd.isna(t):
        return t
    s = t.lower()
    s = re.sub(r'\bsr\.?(?=\s)', 'senior', s)       # 'sr. data analyst' -> 'senior data analyst'
    s = re.sub(r'\bjr\.?(?=\s)', 'junior', s)
    s = s.replace('ml engineer', 'machine learning engineer')
    s = s.replace('business intelligence analyst', 'bi analyst')
    s = s.replace('data scientist intern', 'data science intern')
    s = re.sub(r'python dev\b', 'python developer', s)
    return official_titles.get(s, 'UNMAPPED: ' + t)

df_clean['job_title'] = df_clean['job_title'].apply(clean_title)

print("Unmapped job titles :", df_clean['job_title'].str.startswith('UNMAPPED').sum())
print("Distinct locations  :", df_clean['location'].nunique(), "->", sorted(df_clean['location'].unique()))
print("Distinct companies  :", df_clean['company'].nunique())
print("Distinct job titles :", df_clean['job_title'].nunique())
```

2.4 Fix 3: dates

Two formats are mixed. We try ISO first; whatever fails, we try DD/MM/YYYY. An explicit format= avoids day/month confusion.

```python
def parse_mixed_date(series):
    iso = pd.to_datetime(series, format='%Y-%m-%d', errors='coerce')
    day_first = pd.to_datetime(series, format='%d/%m/%Y', errors='coerce')
    return iso.fillna(day_first)          # use DD/MM/YYYY result only where ISO failed

df_clean['job_posted_date'] = parse_mixed_date(df_clean['job_posted_date'])
df_clean['application_deadline'] = parse_mixed_date(df_clean['application_deadline'])

print("Posted date  : unparsed =", df_clean['job_posted_date'].isna().sum(), "| raw missing =", df['job_posted_date'].isna().sum())
print("Deadline     : unparsed =", df_clean['application_deadline'].isna().sum(), "| raw missing =", df['application_deadline'].isna().sum())
print("\nPosted range :", df_clean['job_posted_date'].min().date(), "to", df_clean['job_posted_date'].max().date())

# Logical check: a deadline must come AFTER the posting date
both = df_clean.dropna(subset=['application_deadline'])
print("Deadlines on/before posting date:", (both['application_deadline'] <= both['job_posted_date']).sum())

# Time features for EDA
df_clean['posted_year'] = df_clean['job_posted_date'].dt.year
df_clean['posted_month'] = df_clean['job_posted_date'].dt.month
df_clean['posted_year_month'] = df_clean['job_posted_date'].dt.to_period('M').astype(str)
df_clean['days_to_deadline'] = (df_clean['application_deadline'] - df_clean['job_posted_date']).dt.days
```

2.5 Fix 4: skill columns (spacing only, for now)

We fix only the comma spacing. Capitalization and aliases belong to Step 5, where we build a proper skill vocabulary. First, let's see why that step is needed.

```python
skill_cols = ['required_skills', 'preferred_skills', 'programming_languages', 'data_tools',
              'machine_learning_skills', 'database_skills', 'cloud_skills', 'soft_skills']

for col in skill_cols:
    df_clean[col] = df_clean[col].str.replace(r'\s*,\s*', ', ', regex=True)

# Preview of the skill-name problem
tokens = df_clean['required_skills'].dropna().str.split(', ').explode()
print("Distinct skill spellings as written :", tokens.nunique())
print("Distinct after lowercasing          :", tokens.str.lower().nunique())

for pattern in ['python', 'power bi|powerbi|pbi', 'scikit|sklearn', 'postgres', 'sql server|mssql|ms sql']:
    found = tokens[tokens.str.lower().str.contains(pattern)].value_counts()
    print(f"\n{pattern}:  {found.to_dict()}")
```

2.6 Fix 5: remove duplicates

We do this after text cleaning, so that Lahore and lahore can't hide duplicates. There are two types:

Exact duplicates: every column identical.
Near-duplicates: the same job re-posted on another site, with a new job_id. We detect these with a business key (company + title + location + description) and keep the earliest posting.

```python
rows_before = len(df_clean)

# 1. Exact duplicates
n_exact = df_clean.duplicated().sum()
df_clean = df_clean.drop_duplicates()
print("Exact duplicates removed :", n_exact)

# 2. Near-duplicates
business_key = ['company', 'job_title', 'location', 'job_description']
df_clean = df_clean.sort_values('job_posted_date')                    # earliest posting first
is_near_dup = df_clean.duplicated(subset=business_key, keep='first')  # keep the earliest
print("Near-duplicates found    :", is_near_dup.sum())

# Inspect one example pair BEFORE deleting (always look at what you delete!)
example_key = df_clean[is_near_dup].iloc[0][business_key].tolist()
mask = (df_clean[business_key] == example_key).all(axis=1)
display(df_clean.loc[mask, ['job_id', 'company', 'job_title', 'location', 'job_source', 'job_posted_date']])
```

Now delete them:

```python
df_clean = df_clean[~is_near_dup].sort_values('job_id').reset_index(drop=True)

print("Rows before:", rows_before)
print("Rows after :", len(df_clean))
print("Removed    :", rows_before - len(df_clean), f"({(rows_before - len(df_clean)) / rows_before * 100:.2f}%)")
print("job_id unique:", df_clean['job_id'].is_unique)
```

2.7 Fix 6: missing values (the key decision step)

Why not just delete rows with missing values?

```python
share = df_clean.isnull().any(axis=1).mean() * 100
print(f"Rows with at least one missing value: {share:.1f}%")
print(f"If we dropped them, we would lose {share:.1f}% of our data!")
```

```python
# company_size and industry: fill from the same company's most common value
for col in ['company_size', 'industry']:
    company_mode = (df_clean.dropna(subset=[col])
                            .groupby('company')[col]
                            .agg(lambda s: s.mode().iloc[0]))
    df_clean[col] = df_clean[col].fillna(df_clean['company'].map(company_mode))
    df_clean[col] = df_clean[col].fillna('Unknown')

# years_experience: median of the same experience_level
group_median = df_clean.groupby('experience_level')['years_experience'].transform('median')
df_clean['years_experience'] = df_clean['years_experience'].fillna(group_median).round().astype(int)

# simple label fills
df_clean['education_required'] = df_clean['education_required'].fillna('Not Specified')
df_clean['visa_sponsorship'] = df_clean['visa_sponsorship'].fillna('Not Specified')
df_clean['job_source'] = df_clean['job_source'].fillna('Unknown')
for col in ['preferred_skills', 'soft_skills', 'data_tools', 'machine_learning_skills', 'database_skills', 'cloud_skills']:
    df_clean[col] = df_clean[col].fillna('Not Specified')

# salary: keep NaN, add a flag
df_clean['salary_missing'] = df_clean['salary_min'].isna()
df_clean['currency'] = df_clean['currency'].fillna('Unknown')

print("Missing values remaining (only those we deliberately kept):")
remaining = df_clean.isnull().sum()
remaining[remaining > 0]
```

2.8 Fix 7: salary comparability

salary_min and salary_max mix PKR and USD. Step 3 asks about minimum and maximum salary, so we create comparable USD versions using the same exchange rates the dataset used.

```python
fx_rate = {2024: 278, 2025: 281, 2026: 282}          # PKR per USD by posting year
rate = df_clean['posted_year'].map(fx_rate)

for col in ['salary_min', 'salary_max']:
    df_clean[col + '_usd'] = np.where(df_clean['currency'] == 'USD',
                                      df_clean[col],                    # already USD
                                      df_clean[col] / rate).round(0)    # PKR -> USD

# Check against the dataset's own normalized column
mid_usd = (df_clean['salary_min_usd'] + df_clean['salary_max_usd']) / 2
diff = (mid_usd - df_clean['salary_normalized_usd']).abs()
print("Max difference vs salary_normalized_usd:", round(diff.max(), 1), "USD (small = rounding only)")
print("Rows with salary_min_usd > salary_max_usd:", (df_clean['salary_min_usd'] > df_clean['salary_max_usd']).sum())

df_clean[['currency', 'salary_min', 'salary_max', 'salary_min_usd', 'salary_max_usd', 'salary_normalized_usd']].dropna().sample(5, random_state=1)
```

3.3 Skills analysis

Caveat: the skill columns still contain spelling variants (PBI, sklearn, Postgres...). For this EDA we use a small quick alias dictionary so the counts are fair. In Step 5 we will build the proper skill vocabulary. Placeholders (Not Specified, Not Required) are excluded, and each skill is counted once per job.

```python
ALIASES = {
 'python3':'python','python programming':'python','powerbi':'power bi','pbi':'power bi','ms power bi':'power bi',
 'microsoft power bi':'power bi','ms sql':'sql server','mssql':'sql server','microsoft sql server':'sql server',
 'postgres':'postgresql','postgre sql':'postgresql','sklearn':'scikit-learn','scikit learn':'scikit-learn',
 'ml':'machine learning','natural language processing':'nlp','dl':'deep learning','cv':'computer vision',
 'tf':'tensorflow','torch':'pytorch','gcp':'google cloud','google cloud platform':'google cloud',
 'amazon web services':'aws','microsoft azure':'azure','mongo db':'mongodb','k8s':'kubernetes',
 'ms excel':'excel','microsoft excel':'excel','advanced excel':'excel',
 'data viz':'data visualization','data visualisation':'data visualization',
 'statistical analysis':'statistics','stats':'statistics','apache spark':'spark','pyspark':'spark',
 'huggingface':'hugging face','google data studio':'looker studio',
}
PLACEHOLDERS = {'not specified', 'not required'}

def explode_skills(col):
    """One row per (job, skill), with spelling variants merged."""
    s = df[['job_id', col]].copy()
    s['raw'] = s[col].str.split(', ')
    s = s.explode('raw')                                                   # one skill per row
    s['key'] = s['raw'].str.strip().str.lower().replace(ALIASES)           # normalize spelling
    s = s[~s['key'].isin(PLACEHOLDERS)]
    display = s.groupby('key')['raw'].agg(lambda x: x.value_counts().idxmax())   # prettiest spelling
    s['skill'] = s['key'].map(display)
    return s.drop_duplicates(['job_id', 'skill'])                          # once per job

skill_groups = {'programming_languages': 'Programming Languages', 'data_tools': 'Data Tools',
                'machine_learning_skills': 'ML Skills', 'database_skills': 'Database Skills',
                'cloud_skills': 'Cloud Skills', 'soft_skills': 'Soft Skills'}

fig, axes = plt.subplots(2, 3, figsize=(18, 10))
for ax, (col, name) in zip(axes.flat, skill_groups.items()):
    s = explode_skills(col)
    top = s['skill'].value_counts().head(10).sort_values()
    ax.barh(top.index, top.values / len(df) * 100, color='#6a4c93')
    ax.set_title(f'Top 10 {name}', fontweight='bold')
    ax.set_xlabel('% of all job postings'); ax.set_ylabel('Skill')
    print(name, (top.sort_values(ascending=False) / len(df) * 100).round(1).to_dict())
plt.tight_layout(); plt.show()
```

Step 4: NLP on Job Descriptions

What we're doing: turning the free-text job_description column into clean tokens and then into numbers (TF-IDF) that a computer can compare.

Why: Computers can't compare sentences, only numbers. If we convert every description into a vector, then "how well does this job fit my profile" becomes a math question (cosine similarity). That is the engine of Step 6 (matching) and part of Step 7 (ML).

I ran all of this on your cleaned dataset, so the expected outputs are real. The variable is df (the same as df_clean). Run each block in its own Colab cell.

4.2 Build the preprocessing pieces

```python
# ---- 1. Technical-term protection: map every variant to ONE token (multi-word terms joined with _)
TERM_MAP = [
 (r'microsoft power bi|ms power bi|power bi|powerbi|\bpbi\b', 'power_bi'),
 (r'scikit[- ]learn|sklearn', 'scikit_learn'),
 (r'natural language processing|text analytics|\bnlp\b', 'nlp'),
 (r'machine learning|\bml\b', 'machine_learning'),
 (r'deep learning|neural networks?|\bdl\b', 'deep_learning'),
 (r'computer vision|image processing|\bcv\b', 'computer_vision'),
 (r'microsoft sql server|ms sql server|ms sql|mssql|sql server', 'sql_server'),
 (r'postgresql|postgres', 'postgresql'),
 (r'google cloud platform|google cloud|\bgcp\b', 'google_cloud'),
 (r'amazon web services|\baws\b', 'aws'),
 (r'microsoft azure|\bazure\b', 'azure'),
 (r'restful apis?|rest apis?|rest endpoints?', 'rest_api'),
 (r'data visuali[sz]ation|data viz|visual storytelling', 'data_visualization'),
 (r'tensorflow/keras|\btensorflow\b|\btf\b', 'tensorflow'),
 (r'\bpytorch\b', 'pytorch'),
 (r'apache spark|pyspark|\bspark\b', 'spark'),
 (r'extract-transform-load \(etl\)|etl/elt|\betl\b', 'etl'),
 (r'hugging ?face', 'hugging_face'),
 (r'google looker studio|looker studio|google data studio', 'looker_studio'),
 (r'ci/cd', 'cicd'), (r'a/b test(?:s|ing)?', 'ab_testing'),
 (r'statistical (?:analysis|modeling|methods)', 'statistics'),
 (r'python 3|python programming', 'python'),
 (r'structured query language \(sql\)|sql querying', 'sql'),
 (r'advanced excel \(pivot tables, vlookup\)|ms excel|microsoft excel', 'excel'),
 (r'docker containers?', 'docker'), (r'\bgit\b', 'git'),
]
TERM_MAP = [(re.compile(p), repl) for p, repl in TERM_MAP]      # compile once = faster

# Words the lemmatizer must NOT touch (this is what prevents pandas -> panda)
PROTECTED = set(repl for _, repl in TERM_MAP) | {
    'python', 'sql', 'pandas', 'numpy', 'tableau', 'excel', 'docker', 'kubernetes', 'hadoop', 'airflow',
    'kafka', 'mongodb', 'mysql', 'oracle', 'redis', 'keras', 'xgboost', 'django', 'flask', 'fastapi',
    'matplotlib', 'seaborn', 'jupyter', 'linux', 'bigquery', 'snowflake', 'databricks', 'redshift', 'dbt',
    'scala', 'java', 'javascript', 'bash', 'opencv', 'langchain', 'r'}

# ---- 2. Stopwords: standard English + job-posting boilerplate + city names
domain_stop = {'role', 'team', 'join', 'looking', 'hiring', 'seeking', 'candidate', 'ideal', 'will', 'work',
    'working', 'must', 'also', 'well', 'including', 'include', 'strong', 'skill', 'skills', 'experience',
    'required', 'preferred', 'nice', 'plus', 'advantage', 'familiarity', 'exposure', 'knowledge', 'ability',
    'based', 'using', 'use', 'help', 'get', 'new', 'day', 'year', 'years'}

boilerplate_stop = {'essential', 'degree', 'bachelor', 'master', 'phd', 'relevant', 'field', 'related',
    'discipline', 'quantitative', 'mathematics', 'computer', 'technical', 'delivering', 'end', 'professional',
    'hand', 'hands', 'someone', 'follows', 'following', 'per', 'week', 'office', 'schedule', 'flexible', 'medical',
    'insurance', 'coverage', 'annual', 'bonus', 'provident', 'fund', 'leave', 'paid', 'mentoring', 'mentorship',
    'mentor', 'workshop', 'upskilling', 'regular', 'competitive', 'salary', 'learning', 'budget', 'course',
    'certification', 'modern', 'workspace', 'power', 'internet', 'backup', 'friendly', 'fast', 'growing', 'room',
    'grow', 'larger', 'possible', 'offer', 'internship', 'benefit', 'hybrid', 'remote', 'fully', 'reliable',
    'connection', 'availability', 'overlapping', 'hour', 'hold', 'bring', 'value', 'formal', 'demonstrable',
    'portfolio', 'self', 'driven', 'recently', 'completed', 'pursuing', 'currently', 'academic', 'personal',
    'project', 'little', 'limited', 'up', 'owning', 'complex', 'track', 'record', 'colleague', 'junior', 'senior',
    'entry', 'level', 'lead', 'associate', 'need', 'needed', 'expected', 'comfortable', 'real', 'problem',
    'alongside', 'collaborative', 'organisation', 'organization', 'achieved'}

cities = {'lahore', 'karachi', 'islamabad', 'rawalpindi', 'peshawar', 'faisalabad', 'multan', 'sialkot',
          'gujranwala', 'hyderabad', 'quetta', 'abbottabad'}

STOP = set(stopwords.words('english')) | domain_stop | boilerplate_stop | cities

# ---- 3. Boilerplate SENTENCES (benefits, work arrangement) removed with a regex
BOILERPLATE = re.compile(r"(we offer|benefits include|you will join a friendly|we provide a modern|this is a paid internship)[^.]*\.|"
                         r"the role follows a hybrid model[^.]*\.|this role is fully remote[^.]*\.")

tokenizer = RegexpTokenizer(r"[a-z0-9_+#]+")      # keeps _ + # so tokens like power_bi and c++ survive
lemm = WordNetLemmatizer()

def preprocess(text, company=None):
    if company:                                    # remove this posting's own company name (noise for role analysis)
        text = re.sub(re.escape(company), ' ', text, flags=re.IGNORECASE)
    t = BOILERPLATE.sub(' ', text.lower())          # lowercase, then drop benefits/arrangement sentences
    for pattern, repl in TERM_MAP:                  # protect technical terms
        t = pattern.sub(repl, t)
    t = re.sub(r'(?<=\w)\+(?=\s|$)', ' ', t)        # '3+ years' -> '3 years'
    out = []
    for tok in tokenizer.tokenize(t):               # tokenization
        if tok.isdigit():                  continue # drop pure numbers
        if len(tok) < 2 and tok != 'r':    continue # drop 1-letter tokens, but keep the R language
        if tok in STOP:                    continue # stopword removal
        lemma = tok if tok in PROTECTED else lemm.lemmatize(tok)   # lemmatize unless protected
        if lemma in STOP:                  continue # 'days' -> 'day' -> stopword
        out.append(lemma)
    return ' '.join(out)
```

4.5 N-gram analysis

An n-gram is a sequence of n adjacent words. Bigrams (2 words) catch phrases that single words miss, like attention detail or critical thinking.

```python
from sklearn.feature_extraction.text import CountVectorizer

def top_ngrams(texts, n, k):
    vec = CountVectorizer(ngram_range=(n, n), min_df=5)
    X_counts = vec.fit_transform(texts)
    freq = np.asarray(X_counts.sum(axis=0)).ravel()           # total occurrences of each n-gram
    return sorted(zip(vec.get_feature_names_out(), freq), key=lambda x: -x[1])[:k]

print("TOP BIGRAMS")
for phrase, count in top_ngrams(df['processed_description'], 2, 20):
    print(f"{count:6,}  {phrase}")

print("\nTOP TRIGRAMS")
for phrase, count in top_ngrams(df['processed_description'], 3, 10):
    print(f"{count:6,}  {phrase}")
```

4.6 TF-IDF

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(
    max_features=5000,      # keep the 5,000 most useful terms
    ngram_range=(1, 2),     # single words AND two-word phrases
    min_df=5,               # ignore terms appearing in fewer than 5 jobs (typos, noise)
    max_df=0.9,             # ignore terms appearing in more than 90% of jobs (no information)
    sublinear_tf=True       # dampen repeated words: 1 + log(tf) instead of raw count
)
X = tfidf.fit_transform(df['processed_description'])

print("TF-IDF matrix shape:", X.shape)
print(f"Share of non-zero cells: {X.nnz / (X.shape[0] * X.shape[1]) * 100:.2f}%")
```

Why IDF matters, shown on our data

```python
vocab = tfidf.vocabulary_
idf = tfidf.idf_

terms = ['python', 'sql', 'machine_learning', 'power_bi', 'spark', 'nlp', 'pytorch', 'kubernetes', 'computer_vision', 'kpi']
pd.DataFrame({
    'term': terms,
    'idf_weight': [round(idf[vocab[t]], 2) for t in terms],
    '% of jobs containing it': [round((X[:, vocab[t]] > 0).sum() / X.shape[0] * 100, 1) for t in terms],
})
```

What makes one job different from another?

```python
feat = tfidf.get_feature_names_out()

def top_terms(row_index, n=8):
    """Highest-weighted single words for one job (bigrams hidden to keep it readable)."""
    row = X[row_index].toarray().ravel()
    idx = [j for j in row.argsort()[::-1] if ' ' not in feat[j]][:n]
    return [(feat[j], round(row[j], 3)) for j in idx]

de_row = df.index[df['job_title'].str.contains('Data Engineer')][0]
bi_row = df.index[df['job_title'].str.contains('BI Analyst')][0]
print(df.loc[de_row, 'job_title'], '->', top_terms(de_row))
print(df.loc[bi_row, 'job_title'], '->', top_terms(bi_row))
```

5.2 Normalize every skill column

```python
PLACEHOLDERS = {'not specified', 'not required', 'unknown', ''}

def normalize_skill(token):
    """'  sklearn ' -> 'Scikit-learn'. Returns None if the token is not in the vocabulary."""
    return ALIAS_INDEX.get(token.strip().lower())

def parse_skill_cell(cell):
    """'Python, python programming, Sql' -> (['Python', 'SQL'], [])  (good skills, unmapped tokens)"""
    if pd.isna(cell):
        return [], []
    good, bad, seen = [], [], set()
    for tok in str(cell).split(','):                    # split on ',' and strip: handles ',' and ', '
        tok = tok.strip()
        if tok.lower() in PLACEHOLDERS:
            continue                                    # 'Not Specified' is NOT a skill
        skill = normalize_skill(tok)
        if skill is None:
            bad.append(tok)                             # keep a record of anything we failed to map
        elif skill not in seen:                         # duplicates inside one cell are removed
            seen.add(skill)
            good.append(skill)
    return good, bad

skill_cols = ['required_skills', 'preferred_skills', 'programming_languages', 'data_tools',
              'machine_learning_skills', 'database_skills', 'cloud_skills', 'soft_skills']

unmapped = Counter()
for col in skill_cols:
    parsed = df[col].apply(parse_skill_cell)
    df[col + '_std'] = parsed.apply(lambda x: x[0])     # new column holding a Python list of skills
    for _, bad in parsed:
        unmapped.update(bad)

print("Unmapped tokens:", dict(unmapped))
df[['required_skills', 'required_skills_std']].head(3)
```

5.6 Skill extraction from free text

Steps 5.1-5.2 handle lists of skills. A user might also paste a CV paragraph, so we need to find skills inside sentences.

```python
def make_pattern(name):
    escaped = re.escape(name.lower()).replace(r'\ ', r'[\s-]+')            # 'power bi' also matches 'power-bi'
    return rf'(?<![a-z0-9+#_]){escaped}s?(?![a-z0-9+#_])'                  # whole words only, plural allowed

# One compiled regex per skill, covering the canonical name AND all its aliases
PATTERNS = {}
for skill, (cat, aliases) in SKILL_VOCAB.items():
    if skill == 'R':
        continue                                                            # 'R' needs special handling
    PATTERNS[skill] = re.compile('|'.join(make_pattern(n) for n in [skill] + aliases), re.IGNORECASE)

# The language R: a CAPITAL R standing alone ('R, Spark'), but not in 'R&D', 'PR' or 'CRM'
R_PATTERN = re.compile(r'(?<![A-Za-z0-9+#_.&])R(?![A-Za-z0-9+#_&])')

def extract_skills(text, include_soft=False):
    found = [s for s, p in PATTERNS.items()
             if p.search(text) and (include_soft or SKILL_CATEGORY[s] != 'soft')]
    if R_PATTERN.search(text):
        found.append('R')
    return found

demo = "I have 2 years of experience with Python3, sklearn and PBI. I also know MS SQL, k8s, GCP and some natural language processing."
print(extract_skills(demo))
```

How good is the extractor?

We have a rare advantage: ground truth. The job's required_skills + preferred_skills lists say which skills each job has, and the extractor only sees the description text. We compare them:

```python
df['skills_from_text'] = df['job_description'].apply(extract_skills)

truth = df.apply(lambda r: set(r['required_skills_std']) | set(r['preferred_skills_std']), axis=1)
pred = df['skills_from_text'].apply(set)

tp = sum(len(t & p) for t, p in zip(truth, pred))     # found and correct
fp = sum(len(p - t) for t, p in zip(truth, pred))     # found but not in the job's lists
fn = sum(len(t - p) for t, p in zip(truth, pred))     # in the lists but not found in the text

precision = tp / (tp + fp)
recall = tp / (tp + fn)
f1 = 2 * precision * recall / (precision + recall)
print(f"Precision: {precision:.3f} | Recall: {recall:.3f} | F1: {f1:.3f}")

fp_counter = Counter(s for t, p in zip(truth, pred) for s in (p - t))
fn_counter = Counter(s for t, p in zip(truth, pred) for s in (t - p))
print("Most common false positives:", fp_counter.most_common(6))
print("Most common misses:", fn_counter.most_common(6))
```

6.2 The matching module

```python
%%writefile /content/AI-Job-Market-Analysis/src/job_matcher.py
"""Job matching engine: skill normalization, 5-component match score, skill gaps."""
import json
import re
from pathlib import Path

import joblib
import numpy as np
import pandas as pd

ROOT = Path(__file__).resolve().parents[1]

LEVELS = ['Internship', 'Entry Level', 'Junior', 'Mid Level', 'Senior']
EDU_ORDER = {'No Degree Required': 0, "Bachelor's": 1, "Bachelor's or Master's": 1, "Master's": 2, 'PhD': 3}
LEVEL_DIFF_SCORE = {0: 1.0, -1: 0.7, -2: 0.4, -3: 0.2, -4: 0.2, 1: 0.5, 2: 0.15, 3: 0.0, 4: 0.0}
WORK_TYPE_SCORE = {('Remote', 'Hybrid'): .5, ('Hybrid', 'Remote'): .5, ('Hybrid', 'On-site'): .5,
                   ('On-site', 'Hybrid'): .5, ('Remote', 'On-site'): .1, ('On-site', 'Remote'): .1}
TWIN_CITIES = {'Islamabad', 'Rawalpindi'}
FINAL_WEIGHTS = dict(skills=0.50, role=0.20, experience=0.15, location=0.10, education=0.05)
FX = {2024: 278, 2025: 281, 2026: 282}          # PKR per USD, same rates as Step 2


class JobMatcher:
    def __init__(self, root=ROOT):
        root = Path(root)
        self.df = pd.read_csv(root / 'data' / 'cleaned_dataset.csv', parse_dates=['job_posted_date', 'application_deadline'])
        for col in ['required_skills_std', 'preferred_skills_std']:
            self.df[col] = self.df[col].fillna('').apply(lambda s: s.split('|') if s else [])
        self.df['base_title'] = self.df['job_title'].str.replace(r'^(Senior|Lead|Junior|Associate) ', '', regex=True)

        self.mlb = joblib.load(root / 'models' / 'skill_binarizer.pkl')
        self.vocab = json.load(open(root / 'models' / 'skill_vocabulary.json'))
        self.skills = list(self.mlb.classes_)
        self.sidx = {s: i for i, s in enumerate(self.skills)}
        self.X_req = self.mlb.transform(self.df['required_skills_std'])
        self.X_pref = self.mlb.transform(self.df['preferred_skills_std'])

        self.n = len(self.df)
        self.idf = np.log((self.n + 1) / (self.X_req.sum(axis=0) + 1)) + 1        # rare skills weigh more

        # alias lookup + regex patterns for free-text extraction
        self.alias_index, self.patterns = {}, {}
        for skill, info in self.vocab.items():
            names = [skill] + info['aliases']
            for name in names:
                self.alias_index[name.lower()] = skill
            if skill != 'R':
                self.patterns[skill] = re.compile('|'.join(self._pattern(nm) for nm in names), re.IGNORECASE)
        self.r_pattern = re.compile(r'(?<![A-Za-z0-9+#_.&])R(?![A-Za-z0-9+#_&])')

        # role tables
        self.role_to_cat = self.df.groupby('base_title')['target_role_category'].agg(lambda s: s.mode().iloc[0]).to_dict()
        self.roles = sorted(self.role_to_cat)
        self.categories = sorted(self.df['target_role_category'].unique())
        centroid = np.array([self.X_req[(self.df['target_role_category'] == c).values].mean(axis=0) for c in self.categories])
        centroid = centroid / np.linalg.norm(centroid, axis=1, keepdims=True)
        self.cat_sim = pd.DataFrame(centroid @ centroid.T, index=self.categories, columns=self.categories)

        # job arrays used by the scorers
        self.level_idx = self.df['experience_level'].map({l: i for i, l in enumerate(LEVELS)}).values
        self.job_years = self.df['years_experience'].values
        self.job_edu = self.df['education_required'].map(EDU_ORDER).values.astype(float)
        self.job_loc = self.df['location'].values
        self.job_wt = self.df['work_type'].values
        self.job_base = self.df['base_title'].values
        self.job_cat = self.df['target_role_category'].values
        rate = self.df['posted_year'].map(FX)
        self.df['salary_max_monthly_pkr'] = self.df['salary_max_usd'] * rate / 12

    # ------------------------------------------------------------ skills
    @staticmethod
    def _pattern(name):
        escaped = re.escape(name.lower()).replace(r'\ ', r'[\s-]+')
        return rf'(?<![a-z0-9+#_]){escaped}s?(?![a-z0-9+#_])'

    def normalize_skill(self, token):
        return self.alias_index.get(token.strip().lower())

    def parse_user_skills(self, text_or_list):
        """Accepts a list, a comma/newline separated string, or a free-text paragraph.
        Returns (recognized_skills, unrecognized_tokens)."""
        if isinstance(text_or_list, (list, tuple, set)):
            tokens, text = list(text_or_list), ' , '.join(text_or_list)
        else:
            text = str(text_or_list)
            tokens = [t for t in re.split(r'[,\n;]+', text) if t.strip()]
        found, unknown = [], []
        for tok in tokens:
            skill = self.normalize_skill(tok)
            if skill:
                found.append(skill)
            elif len(tok.split()) <= 3:
                unknown.append(tok.strip())
        for skill, pattern in self.patterns.items():              # free-text extraction
            if pattern.search(text):
                found.append(skill)
        if self.r_pattern.search(text):
            found.append('R')
        found = list(dict.fromkeys(found))
        unknown = [u for u in unknown if not any(p.search(u) for p in self.patterns.values())]
        return found, unknown

    # ------------------------------------------------------------ components
    def skill_score(self, user_skills):
        u = np.zeros(len(self.skills))
        for s in user_skills:
            if s in self.sidx:
                u[self.sidx[s]] = 1
        req_cov = (self.X_req * self.idf) @ u / (self.X_req @ self.idf)
        pref_w = self.X_pref @ self.idf
        pref_cov = np.divide((self.X_pref * self.idf) @ u, pref_w, out=np.zeros(self.n), where=pref_w > 0)
        return np.where(pref_w > 0, 0.8 * req_cov + 0.2 * pref_cov, req_cov), req_cov

    @staticmethod
    def user_level_index(years):
        return 1 if years < 1 else 2 if years < 3 else 3 if years < 6 else 4

    def experience_score(self, years):
        years_fit = np.where(years >= self.job_years, 1.0, np.clip(1 - 0.35 * (self.job_years - years), 0, 1))
        diff = self.level_idx - self.user_level_index(years)
        level_fit = np.array([LEVEL_DIFF_SCORE[d] for d in diff])
        return 0.6 * years_fit + 0.4 * level_fit

    def education_score(self, education):
        u = EDU_ORDER[education]
        score = np.where(u >= self.job_edu, 1.0, np.where(self.job_edu - u == 1, 0.5, 0.0))
        return np.where(np.isnan(self.job_edu), 0.9, score)              # 'Not Specified' -> neutral

    def role_score(self, target_role=None, target_category=None):
        cat = self.role_to_cat.get(target_role) or target_category
        if cat is None:
            return np.full(self.n, 0.5)                                  # no preference: same for every job
        related = 0.5 * self.cat_sim.loc[cat, self.job_cat].values
        if target_role in self.role_to_cat:
            return np.where(self.job_base == target_role, 1.0, np.where(self.job_cat == cat, 0.7, related))
        return np.where(self.job_cat == cat, 1.0, related)

    def location_score(self, city, work_type):
        if city in (None, 'Any', 'Pakistan'):
            city_fit = np.ones(self.n)
        else:
            twin = (city in TWIN_CITIES) & np.isin(self.job_loc, list(TWIN_CITIES))
            city_fit = np.where(self.job_loc == city, 1.0, np.where(self.job_loc == 'Remote', 1.0, np.where(twin, 0.8, 0.2)))
        if work_type in (None, 'Any'):
            wt_fit = np.ones(self.n)
        else:
            wt_fit = np.array([1.0 if work_type == j else WORK_TYPE_SCORE[(work_type, j)] for j in self.job_wt])
        return 0.5 * city_fit + 0.5 * wt_fit

    # ------------------------------------------------------------ overall
    def score_all(self, user, weights=FINAL_WEIGHTS):
        skill, req_cov = self.skill_score(user['skills'])
        comp = pd.DataFrame({
            'skills_match': skill,
            'role_match': self.role_score(user.get('target_role'), user.get('target_category')),
            'experience_match': self.experience_score(user['experience_years']),
            'location_match': self.location_score(user.get('location'), user.get('work_type')),
            'education_match': self.education_score(user['education']),
        }, index=self.df.index)
        comp['match_score'] = 100 * sum(weights[k] * comp[k + '_match'] for k in weights)
        comp['required_coverage'] = req_cov
        return comp
```

6.5 Choosing the weights with evidence

We have no click data, so we can't learn weights. Instead we simulate 150 users from real jobs (they know 50-90% of a job's skills plus 0-2 random popular ones, with experience within +-1 year) and check which weights behave sensibly.

```python
rng = np.random.default_rng(42)
cities = ['Lahore', 'Karachi', 'Islamabad', 'Rawalpindi', 'Peshawar', 'Faisalabad', 'Multan']
popular = [s for s, _ in Counter(s for l in m.df['required_skills_std'] for s in l).most_common(40)]

def simulate_user(i):
    row = m.df.iloc[i]; req = row['required_skills_std']
    k = max(2, int(round(len(req) * rng.uniform(0.5, 0.9))))
    have = list(rng.choice(req, size=k, replace=False)) + list(rng.choice(popular, size=rng.integers(0, 3), replace=False))
    return dict(skills=list(dict.fromkeys(have)),
                experience_years=max(0, int(row['years_experience']) + int(rng.integers(-1, 2))),
                education=str(rng.choice(["Bachelor's", "Master's", 'PhD'], p=[.65, .3, .05])),
                location=row['location'] if (row['location'] != 'Remote' and rng.random() < .5) else str(rng.choice(cities)),
                work_type=str(rng.choice(['Remote', 'Hybrid', 'On-site', 'Any'], p=[.3, .3, .2, .2])),
                target_role=row['base_title']), i

sim_users = [simulate_user(int(i)) for i in rng.choice(m.n, size=150, replace=False)]

# 1) How much does each component actually separate jobs?
COMPONENTS = ['skills_match', 'role_match', 'experience_match', 'location_match', 'education_match']
spread = pd.DataFrame([m.score_all(u)[COMPONENTS].std() for u, _ in sim_users[:100]]).mean()
weights = pd.Series({k + '_match': w for k, w in FINAL_WEIGHTS.items()})
print(pd.DataFrame({'std_across_jobs': spread.round(3), 'weight': weights, 'influence': (spread * weights).round(3)}))

# 2) Compare weighting schemes
SPEC = dict(skills=.50, experience=.15, education=.10, role=.15, location=.10)
SCHEMES = {'brief: 50/15/10/15/10 (skills/exp/edu/role/loc)': SPEC,
           'final: 50/15/5/20/10': FINAL_WEIGHTS,
           'skills only': dict(skills=1, experience=0, education=0, role=0, location=0),
           'equal 20% each': dict(skills=.2, experience=.2, education=.2, role=.2, location=.2)}
ref_top = [set(m.score_all(u, FINAL_WEIGHTS)['match_score'].nlargest(10).index) for u, _ in sim_users]

rows = []
for name, w in SCHEMES.items():
    overlap, cat_prec, lvl_fit, src = [], [], [], []
    for (u, i), ref in zip(sim_users, ref_top):
        sc = m.score_all(u, w)['match_score']; t10 = sc.nlargest(10).index
        overlap.append(len(set(t10) & ref) / 10)
        cat_prec.append((m.job_cat[t10] == m.job_cat[i]).mean())
        lvl_fit.append((np.abs(m.level_idx[t10] - m.level_idx[i]) <= 1).mean())
        src.append(i in set(sc.nlargest(20).index))
    rows.append((name, np.mean(overlap), np.mean(cat_prec), np.mean(lvl_fit), np.mean(src)))
print(pd.DataFrame(rows, columns=['scheme', 'top10_overlap_with_final', 'same_category@10',
                                  'level_within_1@10', 'source_job_in_top20']).round(3).to_string(index=False))
```

7.3 Features, split, augmentation and pipeline

```python
FEATURES = ['required_skills_std', 'years_experience', 'education_required', 'employment_type']
# Excluded on purpose: job_title, description (leakage); salary (not needed, 8% missing);
# industry/work_type/company_size (tested: no gain); job_id/company/location (not role-related)

X_dev, X_test, y_dev, y_test = train_test_split(df[FEATURES], y, test_size=0.2, stratify=y, random_state=42)
X_train, X_val, y_train, y_val = train_test_split(X_dev, y_dev, test_size=0.2, stratify=y_dev, random_state=42)
print(len(X_train), len(X_val), len(X_test))        # 6304 1576 1970

def keep_fraction(frame, keep, seed):
    """Simulate a user who lists only part of their skills."""
    rng = np.random.default_rng(seed); out = frame.copy()
    def cut(s):
        sk = s.split('|'); k = max(2, int(round(len(sk) * keep)))
        return '|'.join(rng.choice(sk, size=min(k, len(sk)), replace=False))
    out['required_skills_std'] = out['required_skills_std'].apply(cut)
    return out

def augment(X, y, copies=2, seed=0):
    """Training data + extra copies with 30-70% of the skills. Done AFTER splitting, so no leakage."""
    rng = np.random.default_rng(seed); parts_X, parts_y = [X], [y]
    for c in range(copies):
        parts_X.append(keep_fraction(X, rng.uniform(0.3, 0.7), seed + c + 1)); parts_y.append(y)
    return pd.concat(parts_X), pd.concat(parts_y)

def make_pipeline(clf, dense=False):
    prep = ColumnTransformer([
        ('skills', CountVectorizer(token_pattern=r'[^|]+', lowercase=False, binary=True), 'required_skills_std'),
        ('years', StandardScaler(), ['years_experience']),
        ('cats', OneHotEncoder(handle_unknown='ignore'), ['education_required', 'employment_type'])])
    steps = [('prep', prep)]
    if dense:
        steps.append(('dense', FunctionTransformer(lambda X: X.toarray(), accept_sparse=True)))
    return Pipeline(steps + [('clf', clf)])
```

7.4 Compare models on the validation set

```python
CANDIDATES = {
 'Logistic Regression (C=0.1)': (LogisticRegression(C=0.1, max_iter=3000, class_weight='balanced'), False),
 'Logistic Regression (C=1)':   (LogisticRegression(C=1, max_iter=3000, class_weight='balanced'), False),
 'Logistic Regression (C=10)':  (LogisticRegression(C=10, max_iter=3000, class_weight='balanced'), False),
 'Linear SVM':                  (LinearSVC(class_weight='balanced'), False),
 'Random Forest':               (RandomForestClassifier(n_estimators=200, class_weight='balanced_subsample', random_state=42, n_jobs=-1), False),
 'Gradient Boosting':           (HistGradientBoostingClassifier(class_weight='balanced', random_state=42), True)}

def evaluate(model, X, y_true, keep, reps=3):
    accs, f1s = [], []
    for r in range(reps if keep < 1 else 1):                 # average several random skill-drops
        Xk = X if keep == 1 else keep_fraction(X, keep, 100 + r)
        p = model.predict(Xk)
        accs.append(accuracy_score(y_true, p)); f1s.append(f1_score(y_true, p, average='macro'))
    return np.mean(accs), np.mean(f1s)

X_aug, y_aug = augment(X_train, y_train)
print("Training rows:", len(X_train), "-> with augmentation:", len(X_aug))

rows = []
for name, (clf, dense) in CANDIDATES.items():
    t = time.time(); pipe = make_pipeline(clf, dense).fit(X_aug, y_aug); secs = time.time() - t
    res = {k: evaluate(pipe, X_val, y_val, k) for k in (1.0, 0.6, 0.4)}
    rows.append(dict(model=name, **{f'F1@{int(k*100)}%': round(v[1], 3) for k, v in res.items()},
                     **{f'acc@{int(k*100)}%': round(v[0], 3) for k, v in res.items()},
                     avg_macroF1=round(np.mean([v[1] for v in res.values()]), 3), seconds=round(secs, 1)))
val_table = pd.DataFrame(rows).set_index('model')
print(val_table.to_string())

plain = make_pipeline(LogisticRegression(C=1, max_iter=3000, class_weight='balanced')).fit(X_train, y_train)
print("LR WITHOUT augmentation, F1 @100/60/40%:", [round(evaluate(plain, X_val, y_val, k)[1], 3) for k in (1.0, 0.6, 0.4)])
```

7.5 Final model: fit on train+validation, test once

```python
X_dev_aug, y_dev_aug = augment(X_dev, y_dev, seed=5)
final = make_pipeline(LogisticRegression(C=10, max_iter=3000, class_weight='balanced')).fit(X_dev_aug, y_dev_aug)

for k in (1.0, 0.6, 0.4):
    a, f = evaluate(final, X_test, y_test, k)
    print(f"TEST, user lists {int(k*100)}% of skills:  accuracy={a:.3f}  macro-F1={f:.3f}")

X_test60 = keep_fraction(X_test, 0.6, 123)
pred60 = final.predict(X_test60)
print(classification_report(y_test, pred60, digits=3))

labels = sorted(y.unique())
cm = confusion_matrix(y_test, pred60, labels=labels)
plt.figure(figsize=(9, 7))
sns.heatmap(cm / cm.sum(axis=1, keepdims=True), annot=True, fmt='.2f', cmap='Blues', xticklabels=labels, yticklabels=labels)
plt.title('Confusion Matrix (rows = truth, 60% of skills listed)', fontweight='bold')
plt.xlabel('Predicted category'); plt.ylabel('True category')
plt.xticks(rotation=45, ha='right'); plt.tight_layout(); plt.show()
```

STEP 8: Recommendation System

What: combine NLP (free-text skill extraction), ML (role prediction) and the matcher into recommend(profile).                                                                                                          8.1 The recommender module

```python
%%writefile /content/AI-Job-Market-Analysis/src/recommender.py
"""Recommendation system: ML role prediction + match scoring + explanations + skill gaps."""
import sys
from pathlib import Path

import joblib
import numpy as np
import pandas as pd

sys.path.append(str(Path(__file__).resolve().parent))
from job_matcher import JobMatcher, ROOT, FINAL_WEIGHTS   # noqa: E402


class Recommender:
    def __init__(self, root=ROOT):
        self.m = JobMatcher(root)
        self.clf = joblib.load(Path(root) / 'models' / 'role_classifier.pkl')
        self.market_share = pd.Series(self.m.X_req.mean(axis=0), index=self.m.skills)   # % of jobs needing each skill

    # ---------------------------------------------------------------- ML
    def predict_category(self, skills, years, education, employment_type='Full-time'):
        frame = pd.DataFrame([{'required_skills_std': '|'.join(skills), 'years_experience': years,
                               'education_required': education, 'employment_type': employment_type}])
        proba = self.clf.predict_proba(frame)[0]
        return dict(sorted(zip(self.clf.classes_, proba), key=lambda kv: -kv[1]))

    # ---------------------------------------------------------------- formatting helpers
    @staticmethod
    def format_salary(row):
        if pd.isna(row['salary_min']):
            return 'Not disclosed'
        if row['currency'] == 'USD':
            return f"USD {row['salary_min']:,.0f} - {row['salary_max']:,.0f} / year"
        return f"PKR {row['salary_min'] / 12:,.0f} - {row['salary_max'] / 12:,.0f} / month"

    def _why(self, row, user, have, req, role_target, min_salary):
        reasons = [f"You have {len(have)} of {len(req)} required skills"
                   + (f" (e.g. {', '.join(have[:3])})" if have else '')]
        if role_target and row['base_title'] == role_target:
            reasons.append('exact match with your target role')
        elif row['target_role_category'] == user.get('target_category'):
            reasons.append(f"same field ({row['target_role_category']})")
        years = user['experience_years']
        if years >= row['years_experience']:
            reasons.append(f"your {years:g} year(s) meet the {int(row['years_experience'])}-year requirement")
        else:
            reasons.append(f"stretch role: asks {int(row['years_experience'])} years, you have {years:g}")
        if row['work_type'] == 'Remote':
            reasons.append('fully remote')
        elif user.get('location') in (row['location'],):
            reasons.append(f"located in your preferred city ({row['location']})")
        if min_salary and pd.notna(row['salary_max_monthly_pkr']) and row['salary_max_monthly_pkr'] >= min_salary:
            reasons.append('salary range reaches your minimum')
        return '; '.join(reasons) + '.'

    # ---------------------------------------------------------------- main API
    def recommend(self, profile, top_n=10, max_per_company=2, allow_overqualified=False):
        skills, unknown = self.m.parse_user_skills(profile.get('skills', []))
        if not skills:                                       # nothing recognised -> nothing to match
            return pd.DataFrame(), dict(skills=[], unknown_skills=unknown, predicted_category=None,
                                        category_probabilities={}, n_eligible=0)
        probs = self.predict_category(skills, profile['experience_years'], profile['education'],
                                      profile.get('employment_type', 'Full-time')) if skills else {}
        predicted = next(iter(probs), None)
        role_target = profile.get('target_role') if profile.get('target_role') in self.m.role_to_cat else None
        user = dict(profile, skills=skills, target_role=role_target,
                    target_category=self.m.role_to_cat[role_target] if role_target else predicted)

        comp = self.m.score_all(user)
        df = self.m.df
        eligible = comp.index
        min_salary = profile.get('min_salary_pkr_month') or 0
        if min_salary:                                      # unknown salaries are kept (NaN comparison is False)
            eligible = eligible[~(df.loc[eligible, 'salary_max_monthly_pkr'] < min_salary)]
        if not allow_overqualified:                         # hide jobs more than 1 level below the user's level
            user_level = self.m.user_level_index(profile['experience_years'])
            eligible = eligible[self.m.level_idx[eligible] >= user_level - 1]

        picked, per_company = [], {}
        for idx in comp.loc[eligible].sort_values('match_score', ascending=False).index:
            company = df.at[idx, 'company']
            if per_company.get(company, 0) >= max_per_company:
                continue
            per_company[company] = per_company.get(company, 0) + 1
            picked.append(idx)
            if len(picked) == top_n:
                break

        user_set, rows = set(skills), []
        for rank, idx in enumerate(picked, start=1):
            row = df.loc[idx]
            req = row['required_skills_std']
            have = [s for s in req if s in user_set]
            missing = sorted([s for s in req if s not in user_set], key=lambda s: -self.market_share[s])
            rows.append({
                'Rank': rank, 'Job ID': row['job_id'], 'Job Title': row['job_title'], 'Company': row['company'],
                'Location': row['location'], 'Work Type': row['work_type'], 'Experience Level': row['experience_level'],
                'Salary': self.format_salary(row), 'Required Skills': ', '.join(req),
                'Match Score': round(comp.at[idx, 'match_score'], 1),
                'Required Skills Matched': f"{len(have)}/{len(req)} ({len(have) / len(req) * 100:.0f}%)",
                'Skills You Have': ', '.join(have) or '-', 'Missing Skills': ', '.join(missing) or 'None',
                'Why this job is recommended': self._why(row, user, have, req, role_target, min_salary),
                **{c: round(comp.at[idx, c], 2) for c in
                   ['skills_match', 'role_match', 'experience_match', 'location_match', 'education_match']},
            })
        results = pd.DataFrame(rows)
        info = dict(skills=skills, unknown_skills=unknown, predicted_category=predicted,
                    category_probabilities=probs, n_eligible=len(eligible))
        return results, info

    def skill_gap(self, results, user_skills, top_k=8):
        """Aggregate the skills missing across the recommended jobs, weighted by match score."""
        user_set, weight = set(user_skills), {}
        for _, r in results.iterrows():
            for s in (r['Missing Skills'].split(', ') if r['Missing Skills'] != 'None' else []):
                weight[s] = weight.get(s, 0) + r['Match Score']
        table = pd.Series(weight, dtype=float).sort_values(ascending=False).head(top_k)
        out = pd.DataFrame({'skill': table.index,
                            'missing_in_n_jobs': [sum(s in r.split(', ') for r in results['Missing Skills']) for s in table.index],
                            'market_demand_%': [round(self.market_share[s] * 100, 1) for s in table.index]})
        return out
```

```python
%%writefile /content/AI-Job-Market-Analysis/app/app.py
"""AI Job Market Analysis, Skill Extraction, Job Matching and Recommendation System (Streamlit)."""
import sys
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
import streamlit as st

ROOT = Path(__file__).resolve().parents[1]
sys.path.append(str(ROOT / 'src'))
from recommender import Recommender  # noqa: E402

st.set_page_config(page_title='AI Job Market Analyzer', page_icon=':bar_chart:', layout='wide')
sns.set_theme(style='whitegrid')

LEVELS = ['Internship', 'Entry Level', 'Junior', 'Mid Level', 'Senior']
EDUCATION = ["Bachelor's", "Master's", 'PhD', 'No Degree Required']
CITIES = ['Lahore', 'Karachi', 'Islamabad', 'Rawalpindi', 'Peshawar', 'Faisalabad', 'Multan', 'Sialkot',
          'Gujranwala', 'Hyderabad', 'Quetta', 'Abbottabad']


@st.cache_resource(show_spinner='Loading data and models...')
def load_engine():
    return Recommender(ROOT)


engine = load_engine()
jobs = engine.m.df
ROLE_NONE = 'Not sure (let the system decide)'


@st.cache_data
def skill_share(group_col=None, group_value=None, top=None):
    data = jobs if group_col is None else jobs[jobs[group_col] == group_value]
    counts = pd.Series([s for lst in data['required_skills_std'] for s in lst]).value_counts() / len(data) * 100
    return counts.head(top) if top else counts


def tech_skills():
    return sorted(s for s, info in engine.m.vocab.items() if info['category'] != 'soft')


# ------------------------------------------------------------------ pages
def page_dashboard():
    st.title('Dashboard')
    st.caption('Pakistan data and AI job market. The dataset is synthetic (generated for this project).')
    sal = jobs['salary_normalized_usd'].dropna()
    c1, c2, c3, c4, c5 = st.columns(5)
    c1.metric('Total jobs', f'{len(jobs):,}')
    c2.metric('Companies', f"{jobs['company'].nunique():,}")
    c3.metric('Industries', jobs.loc[jobs['industry'] != 'Unknown', 'industry'].nunique())
    c4.metric('Average salary (USD/yr)', f'${sal.mean():,.0f}')
    c5.metric('Median salary (USD/yr)', f'${sal.median():,.0f}')
    left, right = st.columns(2)
    with left:
        st.subheader('Most demanded skills (% of jobs)')
        st.bar_chart(skill_share(top=15).round(1))
    with right:
        st.subheader('Top job categories')
        st.bar_chart(jobs['target_role_category'].value_counts())


def page_search():
    st.title('Job Search')
    c1, c2, c3 = st.columns(3)
    title = c1.text_input('Job title contains', '')
    location = c2.selectbox('Location', ['All'] + sorted(jobs['location'].unique()))
    country = c3.selectbox('Country', ['All'] + sorted(jobs['country'].unique()))
    c4, c5, c6 = st.columns(3)
    levels = c4.multiselect('Experience level', LEVELS)
    work_types = c5.multiselect('Work type', sorted(jobs['work_type'].unique()))
    emp_types = c6.multiselect('Employment type', sorted(jobs['employment_type'].unique()))
    c7, c8 = st.columns(2)
    industries = c7.multiselect('Industry', sorted(jobs['industry'].unique()))
    max_sal = int(jobs['salary_normalized_usd'].max())
    salary = c8.slider('Annual salary (USD)', 0, max_sal, (0, max_sal), step=1000)
    include_unknown = c8.checkbox('Include jobs with undisclosed salary', value=True)

    mask = pd.Series(True, index=jobs.index)
    if title:
        mask &= jobs['job_title'].str.contains(title, case=False, na=False)
    if location != 'All':
        mask &= jobs['location'] == location
    if country != 'All':
        mask &= jobs['country'] == country
    for col, chosen in [('experience_level', levels), ('work_type', work_types),
                        ('employment_type', emp_types), ('industry', industries)]:
        if chosen:
            mask &= jobs[col].isin(chosen)
    in_range = jobs['salary_normalized_usd'].between(*salary)
    mask &= in_range | (jobs['salary_normalized_usd'].isna() & include_unknown)

    result = jobs.loc[mask, ['job_id', 'job_title', 'company', 'location', 'work_type', 'employment_type',
                             'experience_level', 'years_experience', 'industry', 'salary_normalized_usd',
                             'required_skills_std']].copy()
    result['required_skills_std'] = result['required_skills_std'].apply(', '.join)
    result = result.rename(columns={'salary_normalized_usd': 'salary_usd_per_year', 'required_skills_std': 'required_skills'})
    st.write(f'**{len(result):,}** jobs match your filters')
    st.dataframe(result.head(500), hide_index=True)
    st.download_button('Download results (CSV)', result.to_csv(index=False), 'filtered_jobs.csv', 'text/csv')


def profile_inputs(key):
    """Shared profile form used by the Recommendation page."""
    name = st.text_input('Name', 'Ayesha', key=f'{key}_name')
    skills = st.text_area('Your skills (comma separated, or paste a short paragraph about yourself)',
                          'Python, Pandas, NumPy, SQL, Machine Learning', key=f'{key}_skills')
    c1, c2, c3 = st.columns(3)
    education = c1.selectbox('Education', EDUCATION, key=f'{key}_edu')
    years = c2.number_input('Years of experience', 0.0, 15.0, 1.0, 0.5, key=f'{key}_years')
    role = c3.selectbox('Preferred role', [ROLE_NONE] + engine.m.roles, key=f'{key}_role',
                        index=1 + engine.m.roles.index('Data Scientist'))
    c4, c5, c6 = st.columns(3)
    location = c4.selectbox('Preferred location', ['Pakistan (anywhere)'] + CITIES, key=f'{key}_loc')
    work_type = c5.selectbox('Work preference', ['Any', 'Remote', 'Hybrid', 'On-site'], index=1, key=f'{key}_wt')
    min_salary = c6.number_input('Minimum salary (PKR per month, 0 = any)', 0, 2_000_000, 0, 25_000, key=f'{key}_sal')
    return dict(name=name, skills=skills, education=education, experience_years=float(years),
                target_role=None if role == ROLE_NONE else role,
                location='Pakistan' if location.startswith('Pakistan') else location,
                work_type=work_type, min_salary_pkr_month=int(min_salary))


def page_recommend():
    st.title('Job Recommendation')
    profile = profile_inputs('rec')
    if st.button('Find my jobs', type='primary'):
        results, info = engine.recommend(profile, top_n=10)
        if results.empty:
            st.error('We could not recognise any skill in your input. Try names like Python, SQL, Power BI.')
            return
        st.session_state['profile'], st.session_state['results'], st.session_state['info'] = profile, results, info
    if 'results' not in st.session_state:
        return
    results, info, profile = st.session_state['results'], st.session_state['info'], st.session_state['profile']
    st.success(f"Hello {profile['name']}! Recognised skills: {', '.join(info['skills'])}")
    if info['unknown_skills']:
        st.warning('Not recognised (ignored): ' + ', '.join(info['unknown_skills']))
    top3 = list(info['category_probabilities'].items())[:3]
    st.info('Based on your skills, our ML model thinks you fit **' + top3[0][0] + f'** ({top3[0][1] * 100:.0f}% confidence).')
    st.subheader('Top 10 recommended jobs')
    for _, r in results.iterrows():
        with st.expander(f"#{r['Rank']}  {r['Job Title']} at {r['Company']}  |  match {r['Match Score']}%"):
            st.progress(min(int(r['Match Score']), 100))
            a, b, c, d = st.columns(4)
            a.write(f"**Location**\n\n{r['Location']}")
            b.write(f"**Work type**\n\n{r['Work Type']}")
            c.write(f"**Level**\n\n{r['Experience Level']}")
            d.write(f"**Salary**\n\n{r['Salary']}")
            st.write(f"**Required skills:** {r['Required Skills']}")
            st.write(f"**You already have:** {r['Skills You Have']}")
            st.write(f"**Missing skills:** {r['Missing Skills']}")
            st.write(f"**Why:** {r['Why this job is recommended']}")
    st.download_button('Download recommendations (CSV)', results.to_csv(index=False), 'recommendations.csv', 'text/csv')


def page_skill_gap():
    st.title('Skill Gap Analysis')
    default = ', '.join(st.session_state['info']['skills']) if 'info' in st.session_state else 'Python, SQL, Pandas'
    text = st.text_area('Your current skills', default)
    c1, c2 = st.columns(2)
    role = c1.selectbox('Target role', engine.m.roles, index=engine.m.roles.index('Data Scientist'))
    level = c2.selectbox('Level', ['All levels'] + LEVELS)
    threshold = st.slider('A skill counts as "core" when this % of jobs need it', 10, 60, 25)
    if st.button('Analyse my gap', type='primary'):
        have, unknown = engine.m.parse_user_skills(text)
        data = jobs[jobs['base_title'] == role]
        if level != 'All levels':
            data = data[data['experience_level'] == level]
        if data.empty or not have:
            st.error('No jobs for this selection, or no skill recognised.')
            return
        share = pd.Series([s for lst in data['required_skills_std'] for s in lst]).value_counts() / len(data) * 100
        core = share[share >= threshold]
        have_core, missing = core[core.index.isin(have)], core[~core.index.isin(have)]
        match_pct = have_core.sum() / core.sum() * 100
        a, b, c = st.columns(3)
        a.metric('Skill match', f'{match_pct:.0f}%')
        b.metric('Core skills you have', f'{len(have_core)} / {len(core)}')
        c.metric(f'Jobs analysed ({role})', f'{len(data):,}')
        if unknown:
            st.warning('Not recognised (ignored): ' + ', '.join(unknown))
        fig, ax = plt.subplots(figsize=(8, max(3, 0.35 * len(core))))
        order = core.sort_values()
        ax.barh(order.index, order.values, color=['#2a9d8f' if s in have else '#e76f51' for s in order.index])
        ax.set_title(f'Core skills for {role} (green = you have, red = missing)', fontweight='bold')
        ax.set_xlabel('% of jobs requiring the skill'); ax.set_ylabel('Skill')
        st.pyplot(fig)
        col1, col2 = st.columns(2)
        col1.subheader('Current skills'); col1.write(', '.join(have))
        col2.subheader('Missing core skills'); col2.write(', '.join(missing.index) or 'None. You cover every core skill.')
        st.subheader('Recommended skills to learn next')
        st.table(pd.DataFrame({'skill': missing.index, '% of jobs requiring it': missing.round(1).values}).head(8))
        nice = pd.Series([s for lst in data['preferred_skills_std'] for s in lst]).value_counts() / len(data) * 100
        nice = nice[~nice.index.isin(have)].head(5)
        if len(nice):
            st.caption('Nice-to-have skills that could make you stand out: ' + ', '.join(nice.index))


def page_prediction():
    st.title('ML Prediction: Role Category')
    st.write('Enter job or candidate information and the model predicts the **target role category**.')
    skills = st.multiselect('Skills', tech_skills(), default=['Python', 'SQL', 'Machine Learning'])
    c1, c2, c3 = st.columns(3)
    years = c1.number_input('Years of experience', 0, 15, 2)
    education = c2.selectbox('Education', EDUCATION)
    emp = c3.selectbox('Employment type', ['Full-time', 'Part-time', 'Contract', 'Internship'])
    if st.button('Predict', type='primary'):
        if len(skills) < 2:
            st.error('Please select at least 2 skills (the model was trained with at least 2).')
            return
        probs = engine.predict_category(skills, years, education, emp)
        label, top = next(iter(probs.items()))
        st.success(f'Predicted category: **{label}** ({top * 100:.0f}% confidence)')
        st.bar_chart(pd.Series(probs, name='probability'))
        if len(skills) < 4:
            st.caption('Tip: with fewer than 4 skills the model is less accurate (see the README for test results).')


def page_eda():
    st.title('EDA Dashboard')
    tab1, tab2, tab3, tab4 = st.tabs(['Market', 'Salary', 'Skills', 'Trends'])
    with tab1:
        c1, c2 = st.columns(2)
        fig, ax = plt.subplots(figsize=(6, 4))
        jobs['experience_level'].value_counts().reindex(LEVELS).plot.bar(ax=ax, color='#3b6fb6')
        ax.set_title('Jobs by Experience Level', fontweight='bold'); ax.set_xlabel('Experience level'); ax.set_ylabel('Number of jobs')
        c1.pyplot(fig)
        fig, ax = plt.subplots(figsize=(6, 4))
        jobs['work_type'].value_counts().plot.pie(ax=ax, autopct='%1.0f%%', ylabel='')
        ax.set_title('Work Type Share', fontweight='bold')
        c2.pyplot(fig)
        fig, ax = plt.subplots(figsize=(8, 4))
        jobs['location'].value_counts().sort_values().plot.barh(ax=ax, color='#e76f51')
        ax.set_title('Jobs by Location', fontweight='bold'); ax.set_xlabel('Number of jobs'); ax.set_ylabel('City')
        st.pyplot(fig)
    with tab2:
        sal = jobs.dropna(subset=['salary_normalized_usd'])
        fig, ax = plt.subplots(figsize=(8, 4))
        sns.boxplot(data=sal, x='experience_level', y='salary_normalized_usd', order=LEVELS, ax=ax, color='#8ecae6')
        ax.set_title('Annual Salary (USD) by Experience Level', fontweight='bold'); ax.set_xlabel('Experience level'); ax.set_ylabel('Annual salary (USD)')
        st.pyplot(fig)
        med = sal.groupby('target_role_category')['salary_normalized_usd'].median().sort_values()
        fig, ax = plt.subplots(figsize=(8, 4))
        med.plot.barh(ax=ax, color='#2a9d8f')
        ax.set_title('Median Salary (USD) by Role Category', fontweight='bold'); ax.set_xlabel('Median annual salary (USD)'); ax.set_ylabel('Role category')
        st.pyplot(fig)
    with tab3:
        top15 = list(skill_share(top=15).index)
        table = pd.DataFrame({c: (pd.Series([s for lst in g['required_skills_std'] for s in lst]).value_counts() / len(g) * 100)
                              for c, g in jobs.groupby('target_role_category')}).fillna(0).loc[top15]
        fig, ax = plt.subplots(figsize=(9, 6))
        sns.heatmap(table, annot=True, fmt='.0f', cmap='YlGnBu', ax=ax)
        ax.set_title('Required Skills by Role Category (% of jobs)', fontweight='bold'); ax.set_xlabel('Role category'); ax.set_ylabel('Skill')
        st.pyplot(fig)
    with tab4:
        monthly = jobs.groupby('posted_year_month').size()
        fig, ax = plt.subplots(figsize=(9, 4))
        ax.plot(monthly.index, monthly.values, marker='o', color='#3b6fb6')
        ax.set_xticks(range(0, len(monthly), 3)); ax.set_xticklabels(monthly.index[::3], rotation=45)
        ax.set_title('Job Postings per Month', fontweight='bold'); ax.set_xlabel('Month'); ax.set_ylabel('Number of postings')
        st.pyplot(fig)
        st.caption('Postings dated after today are simulated. The dataset is synthetic.')


PAGES = {'Dashboard': page_dashboard, 'Job Search': page_search, 'Job Recommendation': page_recommend,
         'Skill Gap Analysis': page_skill_gap, 'ML Prediction': page_prediction, 'EDA Dashboard': page_eda}
st.sidebar.title('AI Job Market Analyzer')
choice = st.sidebar.radio('Go to', list(PAGES))
st.sidebar.markdown('---')
st.sidebar.caption('MS Data Science FYP. Data is synthetic and for learning purposes.')
PAGES[choice]()
```
